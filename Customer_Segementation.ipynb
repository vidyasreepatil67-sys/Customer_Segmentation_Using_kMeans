Import libraries

```python
import pandas as pd, matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

Customer data

```python
data = {
    'Annual_Income': [15, 16, 17, 18, 19, 70, 72, 75, 78, 80,
                      30, 32, 35, 38, 40, 85, 88, 90, 92, 95],
    'Spending_Score': [20, 22, 18, 25, 21, 80, 85, 78, 82, 88,
                       60, 65, 55, 62, 58, 30, 25, 28, 32, 27]
}
df=pd.DataFrame(data)
print(df)
```

Select features

```python
x = df[['Annual_Income', 'Spending_Score']]
```

Apply K-Means

```python
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
kmeans.fit(x)
labels=kmeans.labels_
centroids=kmeans.cluster_centers_
```

Add cluster to dataframe


```python
df['Cluster'] = labels
print(df)
```

Plot clusters

```python
plt.scatter(x['Annual_Income'],x['Spending_Score'],c=labels)
```

Plot centroids


```python
plt.scatter(centroids[:, 0],centroids[:, 1],color='red',marker='X',s=200)
plt.xlabel("Annual Income")
plt.ylabel("Spending Score")
plt.title("Customer Segmentation using K-Means")
plt.show()
```
